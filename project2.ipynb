```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
import os

os.makedirs("web_scrapping_project", exist_ok=True)
```

```python
URL = "https://realpython.github.io/fake-jobs"

res = requests.get(URL)
```

```python
res.status_code
res.headers.get("Content-type")
```

```python
with open("project2_scraped_data.html","w",encoding="utf-8") as f:
    f.write(res.text)
    print("Content Downloaded...")
```

```python
with open("project2_scraped_data.html","r",encoding="utf-8") as f:
    html_content = f.read()
soup = BeautifulSoup(html_content,"lxml")
all_jobs = soup.select("div.card-content")
```

```python
jobs_data = []
filtered_jobs = []
for job in all_jobs:
    title = job.find("h2").text.strip()

    if "Data" in title or "Python" in title or "ML" in title:
        filtered_jobs.append(job)

for job in filtered_jobs:
    job_title = job.find("h2").text.strip()
    company = job.find("h3").text.strip()
    location = job.select_one(".location").text.strip()
    time = job.find("time").text.strip()
    link_all_content = job.select("a")
    apply_link_content = link_all_content[1]
    apply_link = apply_link_content.get("href")
    jobs_data.append([job_title,company,location,time,apply_link])
    
```

```python
print(jobs_data)
```

```python
df = pd.DataFrame(jobs_data,columns=["Job_title","Company","Location","Posted_Date","Apply_Link"])
df
```
